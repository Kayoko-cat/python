## Định dạng chuỗi với f-string (cách dùng được khuyến khích!)

Cách thanh lịch và phổ biến nhất để chèn giá trị của biến vào trong một chuỗi là dùng **f-string**: thêm ký tự `f` ngay trước dấu ngoặc mở đầu chuỗi, sau đó đặt biến trong cặp dấu ngoặc nhọn `{ }`.

````python
cat = input("What's your cat's name? ")
name = input("What's your name? ")
print(f"hello, my name is{name} and i have a cat named {cat}")
````

## Một số *method* (phương thức) hữu ích của kiểu `str`

Không nên kỳ vọng người dùng luôn nhập liệu gọn gàng, đúng ý ta — vì vậy cần kiểm tra/chuẩn hoá dữ liệu đầu vào. Python cung cấp sẵn nhiều method cho chuỗi:

| Method | Chức năng |
|---|---|
| `.strip()` | Xoá khoảng trắng dư ở đầu và cuối chuỗi |
| `.title()` | Viết hoa chữ cái đầu của mỗi từ |
| `.upper()` | Chuyển toàn bộ thành chữ HOA |
| `.lower()` | Chuyển toàn bộ thành chữ thường |
| `.replace(a, b)` | Thay thế `a` bằng `b` trong chuỗi |

````python
name = input("What's your name? ")
cat = input("What's your cat's name? ")
name = name.strip()     # xoá khoảng trắng dư ở đầu/cuối
name = name.title()     # viết hoa chữ cái đầu mỗi từ
cat = cat.replace("abc", "xyz")  # thay thế chuỗi con "abc" bằng "xyz"
cat = cat.upper()

print(f"hello, my name is {name} and i have a cat named {cat}")
````

Ta có thể **nối các method liên tiếp** (method chaining) để code ngắn gọn, dễ đọc hơn:

````python
name = input("What's your name? ").strip().title()
cat = input("What's your cat's name? ").replace("abc", "xyz").upper()
print(f"hello, my name is {name} and i have a cat named {cat}")
````

2. Chuẩn hoá chuỗi vừa nhập (xoá khoảng trắng dư, viết hoa chữ cái đầu mỗi từ);
3. In ra lời chào và số ký tự của tên (dùng f-string). Ví dụ nếu người dùng nhập `   nguyễn văn an  `, kết quả mong muốn:

### 1.5. Các toán tử số học (arithmetic operators)

| Toán tử | Ý nghĩa | Ví dụ |
|---|---|---|
| `+` | Cộng | `5 + 2` → `7` |
| `-` | Trừ | `5 - 2` → `3` |
| `*` | Nhân | `5 * 2` → `10` |
| `/` | Chia (luôn cho kết quả là số thực `float`) | `5 / 2` → `2.5` |
| `//` | Chia lấy phần nguyên | `5 // 2` → `2` |
| `%` | Chia lấy số dư (modulo) | `5 % 2` → `1` |
| `**` | Lũy thừa | `5 ** 2` → `25` |

> 💡 Toán tử `%` (modulo) rất hữu ích, ví dụ để kiểm tra số chẵn/lẻ, hoặc chia nhóm dữ liệu.

````python
print(5 + 2)
print(5 - 2)
print(5 * 2)
print(5 / 2)
print(5 // 2)
print(5 % 2)
print(5 ** 2)
````

### Làm tròn số với hàm `round()`

Tra tài liệu Python, cú pháp của `round` là `round(number[, ndigits])` — cặp dấu `[ ]` cho biết `ndigits` là tham số tuỳ chọn (optional). Nếu không truyền `ndigits`, kết quả sẽ được làm tròn thành **số nguyên gần nhất**.

````python
x = float(input("What's x? "))
y = float(input("What's y? "))
z = round(x + y)     # làm tròn thành số nguyên gần nhất
print(z)
````

### Định dạng số lớn với dấu phân cách hàng nghìn

Cú pháp f-string `{bien:,}` sẽ tự thêm dấu phẩy để dễ đọc số lớn, ví dụ `1000` sẽ hiển thị thành `1,000`.

````python
x = float(input("What's x? "))
y = float(input("What's y? "))
z = round(x + y)
print(f"{z:,}")
````

## 🚀 Xem trước Chủ đề 2: Tự định nghĩa hàm với `def`

Phần này **không thuộc phạm vi Chủ đề 1**, nhưng để hoàn chỉnh nội dung Lecture 0 (CS50P) và giúp bạn có cái nhìn tổng quan, chúng ta cùng xem nhanh cách tự tạo hàm của riêng mình.

Nếu gọi một hàm chưa được định nghĩa, Python sẽ báo lỗi. Ta tạo hàm mới bằng từ khoá `def`.

⚠️ Lưu ý: Python là ngôn ngữ dùng **thụt đầu dòng (indentation)** để xác định phần thân của hàm/khối lệnh. Mọi dòng thuộc về một hàm phải được thụt vào so với dòng `def ten_ham():`.

````python
def hello():
    print("hello")

name = input("What's your name? ")
hello()
print(name)
````

Ta có thể cho hàm nhận **tham số (parameter)** để linh hoạt hơn, và đặt sẵn **giá trị mặc định (default value)** cho tham số đó:

````python
def hello(to="world"):
    print("hello,", to)

name = input("What's your name? ")
hello(name)     # truyền tên vừa nhập
hello()         # không truyền gì -> dùng giá trị mặc định "world"
````

Một quy ước phổ biến trong Python là tổ chức chương trình với một hàm `main()` — hàm chính, và **gọi `main()` ở cuối file** để khởi động chương trình. Nếu thiếu dòng gọi `main()`, chương trình sẽ không làm gì cả, dù các hàm đã được định nghĩa đầy đủ!

````python
def main():
    name = input("What's your name? ")
    hello(name)
    hello()

def hello(to="world"):
    print("hello,", to)

main()   # <- nếu thiếu dòng này, chương trình sẽ không làm gì cả!
````
