## 2.1. Cấu trúc rẽ nhánh (if – elif – else)

**Cấu trúc điều kiện (conditionals)** cho phép chương trình *ra quyết định* — chọn một trong nhiều "con đường" thực thi, giống như việc chọn đi theo lối bên trái hay lối bên phải.

Python có sẵn các **toán tử so sánh**, luôn cho kết quả là một giá trị `bool` (`True`/`False`):

| Toán tử | Ý nghĩa |
|---|---|
| `>` | Lớn hơn |
| `<` | Nhỏ hơn |
| `>=` | Lớn hơn hoặc bằng |
| `<=` | Nhỏ hơn hoặc bằng |
| `==` | Bằng nhau (**so sánh** — khác với `=` là **gán giá trị**!) |
| `!=` | Khác nhau |

> ⚠️ Lỗi rất thường gặp: nhầm `=` (gán) với `==` (so sánh). Dùng `=` trong câu `if` sẽ bị `SyntaxError`.

### Câu lệnh `if`
Hãy viết chương trình so sánh hai số nguyên `x`, `y`:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x < y:
    print("x is less than y")
````

`if` dùng giá trị `bool` để quyết định có thực thi khối lệnh thụt vào hay không. Nếu `x < y` là `True`, dòng `print` chạy; nếu `False`, chương trình kết thúc mà không in gì.

### Luồng điều khiển (control flow) với nhiều `if` độc lập
Hãy mở rộng để xét đủ 3 trường hợp bằng ba câu `if` **riêng biệt**:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x < y:
    print("x is less than y")
if x > y:
    print("x is greater than y")
if x == y:
    print("x is equal to y")
````

Chương trình vẫn đúng, nhưng **cả ba câu `if` đều luôn được kiểm tra**, dù kết quả câu đầu đã đủ để suy ra hai câu sau. Đây gọi là **luồng điều khiển (control flow)**: Python đánh giá tuần tự từng câu `if` độc lập, không bỏ qua câu nào.

Biểu đồ minh hoạ (mỗi điều kiện đều bị kiểm tra dù trước đó đã có kết quả `True`):

```mermaid
flowchart TD
  A([start]) --> B{x < y}
  B -- True --> C["#quot;x is less than y#quot;"]
  C --> D{x > y}
  D -- True --> E["#quot;x is greater than y#quot;"]
  E --> F{x == y}
  F -- True --> G["#quot;x is equal to y#quot;"]
  G --> H([stop])
  B -- False --> D
  D -- False --> F
  F -- False --> H
```

*(Nếu trình soạn thảo không hiển thị được biểu đồ trên: chương trình luôn phải kiểm tra **cả 3 điều kiện** theo thứ tự, bất kể điều kiện trước đúng hay chưa.)*

👉 Máy tính của bạn có thể không nhận ra sự khác biệt tốc độ, nhưng một máy chủ thực hiện hàng tỷ phép so sánh mỗi ngày sẽ bị ảnh hưởng đáng kể bởi quyết định thiết kế nhỏ này.

### `elif` — giảm số câu hỏi không cần thiết
Nếu điều kiện `if` phía trên đã đúng, các `elif` phía sau **sẽ không được kiểm tra**:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x < y:
    print("x is less than y")
elif x > y:
    print("x is greater than y")
elif x == y:
    print("x is equal to y")
````

### `else` — trường hợp "mặc định" (catch-all)
Nếu `x` không nhỏ hơn **và** không lớn hơn `y`, thì `x` chắc chắn **bằng** `y` — không cần viết hẳn `elif x == y`, dùng `else` để "vét cạn" trường hợp còn lại:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x < y:
    print("x is less than y")
elif x > y:
    print("x is greater than y")
else:
    print("x is equal to y")
````

So với biểu đồ ban đầu, biểu đồ của phiên bản `if/elif/else` đơn giản hơn nhiều — chương trình **nhiều nhất chỉ cần kiểm tra 2 điều kiện**:

```mermaid
flowchart TD
  A([start]) --> B{x < y}
  B -- True --> C["#quot;x is less than y#quot;"]
  B -- False --> D{x > y}
  D -- True --> E["#quot;x is greater than y#quot;"]
  D -- False --> F["#quot;x is equal to y#quot;"]
  F --> G([stop])
  C --> G
  E --> G
```

✅ Đây là ví dụ điển hình về **tái cấu trúc (refactor)** code: vừa đúng, vừa dễ đọc, vừa hiệu quả hơn.

### Toán tử logic `or`
`or` cho phép kiểm tra "một **trong nhiều** điều kiện" có đúng không. Viết lại "x khác y" bằng `or`:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x < y or x > y:
    print("x is not equal to y")
else:
    print("x is equal to y")
````

Kết quả giống nhau, nhưng "x khác y" chính là ý nghĩa của `!=`. Ta chỉ cần hỏi **một câu duy nhất**:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x != y:
    print("x is not equal to y")
else:
    print("x is equal to y")
````

Ngược lại, nếu muốn "trường hợp bằng nhau" là điều kiện chính, ta đảo thứ tự bằng `==`:

````python
x = int(input("What's x? "))
y = int(input("What's y? "))

if x == y:
    print("x is equal to y")
else:
    print("x is not equal to y")
````

Biểu đồ minh hoạ (chỉ một điều kiện duy nhất cần kiểm tra):

```mermaid
flowchart TD
  A([start]) --> B{x == y}
  B -- True --> C["#quot;x is equal to y#quot;"]
  B -- False --> E["#quot;x is not equal to y#quot;"]
  C --> F([stop])
  E --> F
```

> 💡 Luôn dùng đúng **hai** dấu `==` để so sánh. Chỉ gõ một dấu `=` trong `if` sẽ gây lỗi cú pháp.

### Toán tử logic `and`
`and` yêu cầu **tất cả** điều kiện phải đúng. Xây dựng chương trình xếp loại điểm số (thang 0–100):

````python
score = int(input("Score: "))

if score >= 90 and score <= 100:
    print("Grade: A")
elif score >= 80 and score < 90:
    print("Grade: B")
elif score >= 70 and score < 80:
    print("Grade: C")
elif score >= 60 and score < 70:
    print("Grade: D")
else:
    print("Grade: F")
````

Python cho phép **"chuỗi so sánh" (chained comparisons)** — cú pháp khá đặc biệt, giúp code ngắn gọn hơn nhiều so với dùng `and`:

````python
score = int(input("Score: "))

if 90 <= score <= 100:
    print("Grade: A")
elif 80 <= score < 90:
    print("Grade: B")
elif 70 <= score < 80:
    print("Grade: C")
elif 60 <= score < 70:
    print("Grade: D")
else:
    print("Grade: F")
````

Ta còn có thể cải tiến hơn nữa! Vì các điều kiện được kiểm tra **tuần tự if → elif → elif...**, nếu `score >= 90` sai, chương trình đã biết `score < 90` — không cần lặp lại giới hạn trên ở các `elif` sau:

````python
score = int(input("Score: "))

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
````

Chương trình ngắn hơn, **dễ đọc và dễ bảo trì hơn** — vẫn cho kết quả tương đương.

### Toán tử `%` (modulo — chia lấy dư)
**Tính chẵn/lẻ (parity)** kiểm tra được bằng `%`: `4 % 2` → `0` (chia hết); `3 % 2` → `1` (dư 1, không chia hết).

````python
x = int(input("What's x? "))

if x % 2 == 0:
    print("Even")
else:
    print("Odd")
````

### Kết hợp điều kiện với hàm tự định nghĩa
Ở Chủ đề 1, bạn đã biết dùng `def`. Ta tạo hàm riêng kiểm tra chẵn/lẻ, hàm này **trả về** một giá trị `bool`:

````python
def main():
    x = int(input("What's x? "))
    if is_even(x):
        print("Even")
    else:
        print("Odd")


def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False


main()
````

Câu `if is_even(x):` **không có toán tử so sánh nào** — vì `is_even(x)` bản thân đã trả về `True`/`False`. Câu `if` chỉ kiểm tra giá trị `bool` đó.

### "Pythonic" — phong cách viết code đặc trưng của Python
Ta viết gọn `is_even` thành **một dòng**, dùng biểu thức điều kiện (đọc gần giống câu tiếng Anh):

````python
def main():
    x = int(input("What's x? "))
    if is_even(x):
        print("Even")
    else:
        print("Odd")


def is_even(n):
    return True if n % 2 == 0 else False


main()
````

Tối giản hơn nữa: `n % 2 == 0` **chính nó** đã là `True`/`False` — trả về trực tiếp!

````python
def main():
    x = int(input("What's x? "))
    if is_even(x):
        print("Even")
    else:
        print("Odd")


def is_even(n):
    return n % 2 == 0


main()
````

👉 Đây là phiên bản ngắn gọn, dễ đọc, rất "Pythonic".

### Câu lệnh `match`
Tương tự `if/elif/else`, `match` cho phép so khớp một giá trị với nhiều `case`. Xét ví dụ phân loại nhà (Harry Potter) theo tên:

````python
name = input("What's your name? ")

if name == "Harry":
    print("Gryffindor")
elif name == "Hermione":
    print("Gryffindor")
elif name == "Ron":
    print("Gryffindor")
elif name == "Draco":
    print("Slytherin")
else:
    print("Who?")
````

Ba điều kiện đầu cho cùng một kết quả — có thể cải tiến bằng `or`:

````python
name = input("What's your name? ")

if name == "Harry" or name == "Hermione" or name == "Ron":
    print("Gryffindor")
elif name == "Draco":
    print("Slytherin")
else:
    print("Who?")
````

Hoặc dùng `match`/`case`:

````python
name = input("What's your name? ")

match name:
    case "Harry":
        print("Gryffindor")
    case "Hermione":
        print("Gryffindor")
    case "Ron":
        print("Gryffindor")
    case "Draco":
        print("Slytherin")
    case _:
        print("Who?")
````

`match` so sánh giá trị của `name` với từng giá trị sau mỗi `case`. Khi khớp, khối lệnh tương ứng chạy, rồi dừng việc so khớp. Dấu `_` ở `case` cuối khớp với **bất kỳ** giá trị nào — đóng vai trò như `else`.

Cải tiến với dấu `|` (giống `or`, so khớp nhiều giá trị trong một `case`):

````python
name = input("What's your name? ")

match name:
    case "Harry" | "Hermione" | "Ron":
        print("Gryffindor")
    case "Draco":
        print("Slytherin")
    case _:
        print("Who?")
````

**✏️ Bài tập 4 — Phân loại ngày trong tuần bằng `match`:** Viết chương trình:
1. Hỏi người dùng nhập một ngày trong tuần (tiếng Anh viết thường: `monday`, `tuesday`,...).
2. Dùng `match/case` (kết hợp `|` khi cần) để in ra:
   - `"Ngày làm việc"` nếu là `monday` đến `friday`
   - `"Cuối tuần"` nếu là `saturday` hoặc `sunday`
   - `"Không hợp lệ"` với giá trị khác (`case _`)

💡 *Gợi ý:* chuẩn hoá chuỗi nhập vào bằng `.strip().lower()` trước khi `match`, như đã học ở Chủ đề 1.

````python
# TODO: Viết chương trình phân loại ngày trong tuần bằng match/case
day = input("Nhập một ngày trong tuần: ").strip().lower()
match day:
    case "monday"|"tuesday"|"wednesday"|"thursday"|"friday":
        print("Ngày làm việc")
    case "saturday"|"sunday":
        print("cuối tuần")
    case _:
        print("không hợp lệ")
````

## 2.2. Vòng lặp (`for`, `while`)

> ℹ️ **Nguồn tham khảo bổ sung:** Phần này biên soạn theo *Lecture 2 – "Loops"* của CS50P (https://cs50.harvard.edu/python/notes/2/), tiếp nối trực tiếp sau *Lecture 1 – Conditionals*, nhằm đầy đủ nội dung mục 2.2 theo đề cương.

**Vòng lặp (loop)** giúp thực hiện một khối lệnh **lặp đi lặp lại**, thay vì viết `print` nhiều lần thủ công:

````python
# Cách "thủ công" (kém hiệu quả) - nếu cần lặp 100 lần thì sao?
print("meow")
print("meow")
print("meow")
````

### Vòng lặp `while`
`while` gần như có trong **mọi** ngôn ngữ lập trình — lặp lại khối lệnh **miễn là (while)** điều kiện còn đúng:

````python
i = 0
while i < 3:
    print("meow")
    i = i + 1
````

Biến `i` là "bộ đếm" (counter), bắt đầu từ `0`. Mỗi lần lặp: in `"meow"`, rồi tăng `i` lên 1. Vòng lặp dừng ngay khi `i < 3` không còn đúng (khi `i == 3`).

> ⚠️ Nếu quên dòng `i = i + 1`, điều kiện `i < 3` sẽ **luôn đúng** → chương trình chạy **vô hạn (infinite loop)**! Đây là lỗi rất thường gặp với `while`.

### Vòng lặp `for` với `range`
Python có cách viết ngắn gọn hơn để lặp một số lần cố định:

````python
for i in range(3):
    print("meow")
````

`range(3)` sinh dãy 3 giá trị: `0, 1, 2` (bắt đầu từ `0`, không gồm `3`). Mỗi giá trị được gán vào `i`, khối lệnh chạy — in `"meow"` đúng 3 lần.

Ở đây, ta có biến `i` nhưng **không dùng** giá trị của nó ở bất kỳ đâu — nó chỉ đóng vai trò "đếm số lần lặp". Theo quy ước **Pythonic**, khi biến không có ý nghĩa sử dụng cụ thể, ta đặt tên là `_`:

````python
for _ in range(3):
    print("meow")
````

Kết quả giống nhau, nhưng cách viết này giúp người đọc hiểu ngay: "*giá trị của biến lặp không quan trọng, chỉ cần lặp 3 lần*".

### `continue` và `break`
Muốn hỏi người dùng **cho đến khi** họ nhập một số hợp lệ (không âm), kết hợp `while True` với `continue`/`break`:

````python
while True:
    n = int(input("What's n? "))
    if n < 0:
        continue
    else:
        break

print(f"Thanks for {n}!")
````

Hai từ khoá mới:
- **`continue`**: bỏ qua phần còn lại của lần lặp hiện tại, **quay lại ngay đầu vòng lặp** (ở đây: hỏi lại khi `n < 0`).
- **`break`**: **thoát ngay** khỏi vòng lặp (ở đây: khi `n` đã hợp lệ).

Hoá ra `continue` ở đây là **thừa**! Nếu `n >= 0` thì `break`; nếu không, `while True` sẽ **tự động** lặp lại từ đầu:

````python
while True:
    n = int(input("What's n? "))
    if n >= 0:
        break

print(f"Thanks for {n}!")
````

Ta có thể kết hợp mẫu này với vòng lặp `for` ở trên để tạo chương trình hoàn chỉnh: hỏi người dùng số lần muốn "meow" (phải **dương**, vì meow 0 lần thì vô nghĩa), rồi lặp in ra đúng số lần đó:

````python
while True:
    n = int(input("What's n? "))
    if n > 0:
        break

for _ in range(n):
    print("meow")
````

👉 Đây là ví dụ đẹp về việc **kết hợp** `while` (kiểm tra/nhắc lại đầu vào cho đến khi hợp lệ) với `for` (lặp một số lần xác định) trong cùng một chương trình.

**✏️ Bài tập 5 — Vòng lặp xác nhận:** Viết chương trình dùng `while` để liên tục hỏi: `"Bạn đồng ý với các điều khoản không? (yes/no): "` cho đến khi người dùng nhập chính xác `"yes"` hoặc `"no"` (không phân biệt hoa/thường — gợi ý dùng `.strip().lower()`). Sau khi hợp lệ, in ra:
- `"Cảm ơn bạn đã đồng ý!"` nếu nhập `yes`
- `"Rất tiếc, bạn cần đồng ý để tiếp tục."` nếu nhập `no`

💡 *Gợi ý:* dùng `while True:` kết hợp `break` khi giá trị hợp lệ, sau đó `if/else` để in phản hồi tương ứng.

````python
# TODO: Viết vòng lặp xác nhận yes/no
while True:
    answer = input("Bạn có đồng ý với các điều khoản không? (yes/no): ").strip().lower()
    if answer == "yes":
        print("Cảm ơn bạn đã đồng ý")
        break
    elif answer == "no":
        print("Rất tiếc, bạn cần đồng ý để tiếp tục")
        break
    else:
        print("Vui lòng trả lời 'yes' hoặc 'no'.")
````

### Dùng `for` để lặp qua từng ký tự của một chuỗi
`for` còn lặp trực tiếp qua từng ký tự của một `str` — hữu ích khi xử lý văn bản (kỹ năng quan trọng khi xử lý dữ liệu cho học máy!):

````python
word = input("Enter a word: ")
count = 0

for letter in word:
    if letter.lower() in "aeiou":
        count = count + 1

print(f"Số nguyên âm: {count}")
````

`letter` lần lượt nhận giá trị từng ký tự trong `word`; với mỗi ký tự, kiểm tra (sau khi `.lower()`) có thuộc `"aeiou"` không, nếu có thì tăng `count`.

**✏️ Bài tập 6 — Tính tổng từ 1 đến n:** Viết chương trình:
1. Hỏi người dùng nhập số nguyên dương `n`.
2. Dùng `for` (kết hợp `range`) để tính **tổng** các số nguyên từ `1` đến `n` (bao gồm `n`).
3. In kết quả. Ví dụ `n = 5` → tổng `= 15`.

💡 *Gợi ý:* khởi tạo `tong = 0` trước vòng lặp; `range(n)` chỉ chạy từ `0` đến `n-1`, bạn cần `range(1, n + 1)` để lặp từ `1` đến `n`.

````python
# TODO: Viết chương trình tính tổng các số từ 1 đến n
n = int(input("Nhập một số nguyên dương n: "))
total = 0
for i in range(1, n + 1):
    total += i
print(f"Tổng các số từ 1 đến {n} là: {total}")
````

## 2.3. Hàm và đối số (Functions and Arguments)

Ở Chủ đề 1, bạn đã biết định nghĩa hàm bằng `def`, dùng tham số, giá trị mặc định và `return`. Ví dụ `is_even` ở Phần 2.1 cho thấy: **hàm kết hợp với `if/else`, nhận một đối số, trả về `bool`**. Phần này mở rộng thêm về **đối số (arguments)**.

### Đối số vị trí (positional) và đối số từ khoá (keyword)
Một hàm có thể nhận nhiều tham số. Mặc định, Python gán giá trị theo **đúng vị trí (thứ tự)** — gọi là **đối số vị trí**:

````python
def area(width, height):
    return width * height


print(area(3, 4))     # width=3, height=4 -> 12
````

Ta cũng có thể gọi hàm bằng cách **chỉ rõ tên tham số** — gọi là **đối số từ khoá (keyword argument)**. Khi đó, thứ tự truyền vào **không còn quan trọng**:

````python
def area(width, height):
    return width * height


print(area(height=4, width=3))   # kết quả vẫn là 12, dù đổi thứ tự
print(area(width=5, height=2))
````

### Giá trị mặc định của tham số (nhắc lại + mở rộng)
Nhắc lại từ Chủ đề 1: có thể gán sẵn giá trị mặc định cho tham số. Nếu không truyền giá trị, Python dùng giá trị mặc định:

````python
def area(width, height=2):
    return width * height


print(area(5))          # height dùng giá trị mặc định 2 -> 10
print(area(5, 10))      # height được truyền = 10 -> 50
````

> ⚠️ Tham số **có** giá trị mặc định phải luôn khai báo **sau** tham số **không** có giá trị mặc định (`def area(width, height=2)` hợp lệ, nhưng `def area(width=2, height)` sẽ báo lỗi).

### Hàm gọi hàm khác + kết hợp với điều kiện
Giống hàm `square` ở Chủ đề 1, một hàm có thể **gọi hàm khác**. Hãy viết hàm phức tạp hơn: `is_leap_year(year)` — kiểm tra năm nhuận, kết hợp `and`/`or` theo phong cách Pythonic:

````python
def main():
    year = int(input("Year: "))
    if is_leap_year(year):
        print(f"{year} là năm nhuận")
    else:
        print(f"{year} không là năm nhuận")


def is_leap_year(year):
    # Năm nhuận: chia hết cho 4, trừ khi chia hết cho 100 (nhưng không chia hết cho 400)
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


main()
````

Dù quy tắc năm nhuận khá phức tạp, hàm `is_leap_year` vẫn viết **gọn trong một dòng `return`** — Pythonic, dễ đọc, dễ tái sử dụng ở bất kỳ đâu trong chương trình.

**✏️ Bài tập 7 — Hàm phân loại tam giác:** Viết hàm `classify_triangle(a, b, c)` nhận độ dài 3 cạnh (số thực), trả về chuỗi mô tả loại tam giác, theo **thứ tự kiểm tra** (để tối ưu, giống nguyên tắc `if/elif/else` đã học):
1. Nếu 3 cạnh **không** tạo thành tam giác hợp lệ → `"Không phải tam giác"`
2. Nếu cả 3 cạnh bằng nhau → `"Tam giác đều"`
3. Nếu có đúng 2 cạnh bằng nhau → `"Tam giác cân"`
4. Ngược lại → `"Tam giác thường"`

Viết `main()` hỏi người dùng nhập 3 cạnh và in kết quả gọi từ `classify_triangle`. Nhớ gọi `main()` ở cuối file!

💡 *Gợi ý:* điều kiện tam giác hợp lệ: `a + b > c and a + c > b and b + c > a`.

````python
# TODO: Viết hàm classify_triangle(a, b, c) và hàm main() sử dụng nó
def classify_triangle(a, b, c):
    if a == b == c:
        return "Tam giác đều"
    elif a == b or b == c or a == c:
        return "Tam giác cân"
    else:
        return "Tam giác thường"

def main():
    a = float(input("Nhập cạnh a: "))
    b = float(input("Nhập cạnh b: "))
    c = float(input("Nhập cạnh c: "))
    print(classify_triangle(a, b, c))
main()
````

## 2.4. Phạm vi biến (Variable Scope)

> ℹ️ **Nguồn tham khảo bổ sung:** Khái niệm biến cục bộ được dùng ngầm trong mọi hàm từ Chủ đề 1; phần "biến toàn cục / từ khoá `global`" biên soạn theo *Lecture 9 – "Et Cetera"* của CS50P (https://cs50.harvard.edu/python/notes/9/), đưa vào đây sớm hơn để bám đúng mục 2.4 của đề cương.

**Phạm vi biến (scope)** xác định "vùng" trong chương trình mà một biến có thể được truy cập. Quan sát đoạn code sau (**có lỗi có chủ đích**) và dự đoán điều gì xảy ra:

````python
# ⚠️ Đoạn code này có lỗi CÓ CHỦ ĐÍCH - hãy chạy để quan sát lỗi NameError
def main():
    x = 1
    print_x()


def print_x():
    print(x)


main()
````

Bạn nhận được lỗi `NameError: name 'x' is not defined`! Dù `x` đã được tạo bên trong `main()`, hàm `print_x()` **không thể "nhìn thấy"** biến đó.

👉 Lý do: `x` được tạo **bên trong** `main()` là một **biến cục bộ (local variable)** — nó **chỉ tồn tại và chỉ truy cập được bên trong** hàm khai báo nó. Khi hàm kết thúc, biến cục bộ "biến mất". Các hàm khác hoàn toàn không biết đến sự tồn tại của nó.

### Biến toàn cục (global variable)
Một biến khai báo **bên ngoài mọi hàm** (cấp cao nhất của file) là **biến toàn cục** — mọi hàm đều có thể **đọc (read)** được:

````python
balance = 0


def main():
    print("Balance:", balance)


main()
````

Chạy đúng — vì `balance` là biến toàn cục, hàm `main()` **đọc** được giá trị (`Balance: 0`) mà không cần truyền qua tham số.

### ⚠️ Cạm bẫy thường gặp: gán lại biến toàn cục bên trong hàm
Điều gì xảy ra nếu ta thử **thay đổi** giá trị biến toàn cục ngay trong một hàm?

````python
# ⚠️ Đoạn code này có lỗi CÓ CHỦ ĐÍCH - hãy chạy để quan sát lỗi UnboundLocalError
balance = 0


def main():
    deposit(100)
    print("Balance:", balance)


def deposit(n):
    balance = balance + n


main()
````

Lỗi `UnboundLocalError`! Ngay khi Python thấy dòng `balance = balance + n` bên trong `deposit`, nó hiểu bạn muốn tạo một biến `balance` **cục bộ, mới, riêng** cho hàm `deposit` — và vì biến cục bộ này **chưa có giá trị** trước khi được gán, Python báo lỗi khi cố đọc nó ở bên phải dấu `=`.

### Từ khoá `global` — cách sửa đổi biến toàn cục đúng cách
Để hàm có thể **thay đổi** biến toàn cục, cần khai báo rõ với `global`:

````python
balance = 0


def main():
    print("Balance:", balance)
    deposit(100)
    withdraw(50)
    print("Balance:", balance)


def deposit(n):
    global balance
    balance += n


def withdraw(n):
    global balance
    balance -= n


main()
````

Nhờ `global balance` bên trong mỗi hàm, Python hiểu `balance` dùng trong hàm đó **chính là** biến toàn cục ở đầu file — không phải biến cục bộ mới.

> 💡 **Lời khuyên quan trọng:** Dù `global` giúp chương trình chạy đúng, các lập trình viên (và CS50!) khuyến cáo nên **hạn chế tối đa** việc dùng biến toàn cục và `global`. Cách viết "chuẩn mực" hơn là **truyền dữ liệu vào hàm qua tham số**, và **lấy dữ liệu ra qua `return`** — giúp hàm trở nên độc lập, dễ kiểm tra (test) và tái sử dụng, rất quan trọng khi viết code cho các dự án lớn (kể cả trong học máy!). Hãy chạy thử phiên bản viết lại không dùng `global` dưới đây — kết quả hoàn toàn tương đương:

````python
def main():
    balance = 0
    balance = deposit(balance, 100)
    balance = withdraw(balance, 50)
    print("Balance:", balance)


def deposit(balance, n):
    return balance + n


def withdraw(balance, n):
    return balance - n


main()
````

**✏️ Bài tập 8 — Dự đoán và sửa lỗi phạm vi biến:** Chạy đoạn code có lỗi có chủ đích dưới đây, quan sát lỗi xảy ra, rồi ở ô kế tiếp: (1) giải thích bằng comment `#` **tại sao** có lỗi đó; (2) **sửa lại** code (dùng `global`, HOẶC — tốt hơn — dùng tham số + `return` như khuyến cáo trên) để chương trình in đúng `Tổng: 30`.

````python
# ⚠️ Đoạn code này có lỗi CÓ CHỦ ĐÍCH - hãy chạy để quan sát lỗi
total = 0

def main():
    add_to_total(10)
    add_to_total(20)
    print("Tổng:", total)

def add_to_total(n):
    total = total + n

main()
````

## 🏁 Bài tập tổng hợp cuối Chủ đề 2 — "FizzBuzz"

**FizzBuzz** là bài tập lập trình rất nổi tiếng, kiểm tra khả năng kết hợp **vòng lặp + toán tử modulo + cấu trúc rẽ nhánh** — đúng 3 nội dung chính bạn vừa học!

**Yêu cầu:** In ra các số từ `1` đến `30`, với quy tắc:
- Chia hết cho cả `3` **và** `5` → in `FizzBuzz`
- Chỉ chia hết cho `3` → in `Fizz`
- Chỉ chia hết cho `5` → in `Buzz`
- Ngược lại → in chính số đó

Ví dụ 15 số đầu tiên:
```

4
Buzz
Fizz

8

```

💡 *Gợi ý:*
