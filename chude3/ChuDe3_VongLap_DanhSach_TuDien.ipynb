## 3.1. Vòng lặp (Loops): `while` và `for`

Về bản chất, **vòng lặp (loop)** là cách để thực hiện một điều gì đó lặp đi lặp lại nhiều lần.

Hãy tưởng tượng bạn cần "meow" 3 lần:

````python
print("meow")
print("meow")
print("meow")
````

Đoạn code trên chạy đúng, nhưng nếu cần "meow" **500 lần** thì sao? Gõ lại `print("meow")` 500 lần là không thực tế — đây chính là lúc vòng lặp phát huy sức mạnh.

### Vòng lặp `while`
`while` là cấu trúc gần như có mặt trong **mọi** ngôn ngữ lập trình: lặp lại một khối lệnh **miễn là (while)** điều kiện còn đúng.

> ⚠️ **CẢNH BÁO — vòng lặp vô hạn:** Đoạn code dưới đây **chỉ trình bày dưới dạng văn bản minh hoạ** (không phải ô code có thể chạy!) vì nó sẽ khiến chương trình **lặp mãi không dừng** — `i` luôn giữ giá trị `3`, không bao giờ bằng `0`:
> ```python
> i = 3
> while i != 0:
>     print("meow")
> ```
> `while` hoạt động bằng cách liên tục hỏi: "điều kiện của vòng lặp đã thoả chưa?" — cụ thể: "`i` có khác `0` không?". Vì `i` không đổi, câu trả lời luôn là *có* → lặp vô hạn! Nếu vô tình chạy một chương trình như vậy trong terminal, hãy nhấn **Ctrl + C** để buộc dừng.

Để sửa lỗi, ta cần giảm dần `i` sau mỗi lần lặp:

````python
i = 3
while i != 0:
    print("meow")
    i=i-1
````

Sau mỗi lần lặp — gọi là một **"iteration"** (một vòng của chu trình lặp) — `i` giảm 1, vòng lặp dừng khi `i == 0`. Trong lập trình, ta thường **đếm bắt đầu từ 0**: lần lặp đầu tiên là "lần lặp thứ 0" (0th iteration), lần thứ hai là "1st iteration", v.v.

Ta có thể viết lại theo cách đếm quen thuộc với con người (1, 2, 3):

````python
i = 1
while i <= 3:
    print("meow")
    i = i + 1
````

`i = i + 1` nghĩa là: tính giá trị bên phải (`i + 1`) rồi **gán** (từ phải qua trái) vào lại biến `i`. Tuy nhiên, **thông lệ tốt (best practice)** trong lập trình là bắt đầu đếm từ **0**:

````python
i = 0
while i <=  3:
    print("meow")
    i += 1
````

Lưu ý phép so sánh đổi thành `i < 3` (không phải `<=`) để vẫn lặp đúng 3 lần khi bắt đầu từ 0. `i += 1` là cách viết ngắn gọn, **hoàn toàn tương đương** `i = i + 1`.

Sơ đồ minh hoạ luồng chạy:

```mermaid
flowchart TD

  B --> C{i < 3}
  C -- True --> D["#quot;meow#quot;"]
  D --> E[i += 1]
  E --> C
  C -- False --> F([stop])
```

👉 Vòng lặp đếm `i` tăng dần **tới nhưng không vượt qua** `3`.

**✏️ Bài tập 1 — Đếm ngược bằng `while`:** Viết chương trình:
1. Hỏi người dùng nhập một số nguyên dương `n` (giả sử luôn nhập đúng số dương).
2. Dùng `while` để in ra các số từ `n` **giảm dần** về `1` (mỗi số một dòng), rồi in thêm dòng `"Hết giờ!"`.

Ví dụ `n = 3`, kết quả:
```

```

````python
while True:
    n = int(input("nhap n: "))
    if n > 0:
        break  # Nhập đúng số dương thì thoát vòng lặp
    print("khong hop le moi nhap lai")

while 1 <= n:
    print(f"{n}")
    n = n-1
print("Het gio!")
````

### Vòng lặp `for`
`for` là một loại vòng lặp khác. Để hiểu `for`, ta cần một kiểu biến mới: **`list`** (danh sách) — giống "danh sách mua sắm" hay "danh sách công việc" trong đời thực.

`for` lặp qua từng phần tử của một `list`:

````python
for i in [0, 1, 2]:
    print("meow")
````

Ngắn gọn hơn hẳn `while`: `i` lần lượt nhận `0`, rồi `1`, rồi `2` — mỗi lần đều "meow" — rồi kết thúc.

Nhưng nếu cần lặp đến **1 triệu** lần thì viết `list` `[0, 1, 2, ..., 999999]` là không thực tế! Python cung cấp `range()` để tự sinh dãy số:

````python
for i in range(3):
    print("meow")
````

`range(3)` tự động sinh 3 giá trị (`0, 1, 2`) — không cần liệt kê tay.

Nhận thấy `i` **chưa từng được dùng** bên trong khối lệnh — theo quy ước **Pythonic**, khi biến lặp không có ý nghĩa sử dụng riêng, ta đặt tên là `_`:

````python
for _ in range(3):
    print("meow")
````

Đổi `i` thành `_` không ảnh hưởng kết quả, chỉ giúp người đọc hiểu ngay: "giá trị lặp không quan trọng, chỉ cần lặp 3 lần".

Để mở rộng tư duy, hãy xem đoạn code sau:

````python
print("meow" * 3)
````

Chuỗi `"meow"` nhân với `3` lặp lại chính nó, nhưng kết quả `"meowmeowmeow"` dính liền nhau, không có dấu ngắt dòng! Làm sao để mỗi "meow" xuống dòng riêng?

````python
print("meow\n" * 3, end="")
````

`\n` chèn dấu ngắt dòng vào cuối mỗi "meow"; `end=""` báo cho `print` **không** tự thêm dấu ngắt dòng nào nữa sau chuỗi cuối cùng (tránh dư một dòng trống).

**✏️ Bài tập 2 — Bảng cửu chương bằng `for`:** Viết chương trình:
1. Hỏi người dùng nhập một số nguyên `n` (giả sử từ 1 đến 9).
2. Dùng `for i in range(1, 11):` để in bảng cửu chương của `n` từ 1 đến 10, theo mẫu: `n x i = kết_quả` (ví dụ `n=3`: `"3 x 1 = 3"`, ..., `"3 x 10 = 30"`).

💡 *Gợi ý:* `range(1, 11)` sinh dãy `1, 2, ..., 10`. Dùng f-string: `f"{n} x {i} = {n * i}"`.

### Cải thiện với đầu vào của người dùng (Improving with User Input)

Vòng lặp còn thường dùng để **kiểm tra tính hợp lệ (validate)** của dữ liệu nhập vào. Một khuôn mẫu (paradigm) rất phổ biến trong Python là dùng `while` để "hỏi lại" cho đến khi người dùng nhập đúng.

Ví dụ: yêu cầu người dùng nhập một số **lớn hơn hoặc bằng 0**:

````python
while True:
    n = int(input("What's n? "))
    if n < 0:
        continue
    else:
        break

print(f"Bạn đã nhập: {n}")
````

Hai từ khoá mới:
- **`continue`**: yêu cầu Python chuyển ngay sang **lần lặp kế tiếp** (bỏ qua phần code còn lại trong lần lặp hiện tại).
- **`break`**: yêu cầu Python "thoát" khỏi vòng lặp **ngay lập tức**, dù chưa hoàn tất mọi lần lặp dự kiến.

Cụ thể: khi `n < 0`, ta `continue` để hỏi lại. Khi `n >= 0`, ta `break` để thoát vòng lặp.

Hoá ra, `continue` ở đây là **thừa**! Ta viết gọn hơn (lần này yêu cầu số phải **lớn hơn 0**, không chỉ lớn hơn hoặc bằng 0):

````python
while True:
    n = int(input("What's n? "))
    if n > 0:
        break

for _ in range(n):
    print("meow")
````

Nhìn qua, `while True:` có vẻ chạy vô hạn vì `True` luôn đúng. Nhưng khi `n > 0`, `break` thoát ngay khỏi vòng lặp.

Kết hợp kiến thức về hàm, ta cải thiện code hơn nữa:

````python
def main():
    meow(get_number())


def get_number():
    while True:
        n = int(input("What's n? "))
        if n > 0:
            return n


def meow(n):
    for _ in range(n):
        print("meow")


main()
````

Ta không chỉ tách code thành nhiều hàm, mà còn dùng `return` để trả `n` về cho `main`. 💡 Điểm thú vị: `return` ở đây đóng vai trò tương tự `break` — nó thoát ngay khỏi `while` (và cả hàm `get_number`) khi `n > 0` được thoả.

````python
students = ["Hermione", "Harry", "Ron"]

print(students[0])
print(students[1])
print(students[2])
````

Mỗi phần tử được truy cập bằng **chỉ số (index)** trong dấu `[]` — chỉ số bắt đầu từ `0`: `students[0]` là `"Hermione"`, `students[1]` là `"Harry"`, `students[2]` là `"Ron"`.

## 3.3. Độ dài danh sách — hàm `len()`

Ta dùng `len()` để kiểm tra độ dài (số phần tử) của một `list`.

Giả sử ta muốn in cả **vị trí (số thứ tự)** của học sinh, không chỉ tên:

````python
students = ["Hermione", "Harry", "Ron"]

for i in range(len(students)):
    print(i + 1, students[i])
````

`len(students)` trả về số phần tử của `students` (ở đây là `3`), nên `range(len(students))` sinh ra các chỉ số `0, 1, 2` — đúng bằng các chỉ số hợp lệ của `students`. Với mỗi chỉ số `i`, ta in `i + 1` (số thứ tự dễ đọc, bắt đầu từ 1) và `students[i]` (tên tương ứng).

💡 Điểm mạnh của `len()`: nó luôn tính đúng độ dài **hiện tại** của `list`, dù `list` có bao nhiêu phần tử — không cần biết trước hay "hard-code" con số `3`.

> 📚 Tìm hiểu thêm: [Python's documentation on `len`](https://docs.python.org/3/library/functions.html#len)

**✏️ Bài tập 5 — Tính trung bình danh sách điểm số:** Viết chương trình:
1. Tạo `list` `scores` chứa ít nhất 5 điểm số, ví dụ: `[8.5, 7, 9, 6.5, 10]`.
2. Dùng `for` kết hợp `len()` để tính **tổng** các điểm (không dùng hàm `sum()` có sẵn — tự viết vòng lặp cộng dồn!).
3. Tính điểm trung bình = tổng / `len(scores)`, rồi in kết quả.

💡 *Gợi ý:* có thể lặp trực tiếp `for score in scores:` (không cần `range`/index) vì ta không cần biết vị trí, chỉ cần giá trị.

````python
# TODO: Tính điểm trung bình từ list scores


# 1. Tạo list scores chứa ít nhất 5 điểm số
scores = [8.5, 7, 9, 6.5, 10]

# 2. Dùng vòng lặp for để tính tổng (không dùng hàm sum)
total = 0
for score in scores:
    total += score

# 3. Tính điểm trung bình và in kết quả
average = total / len(scores)


print(f"Điểm trung bình: {average}")
````

## 3.4. Từ điển (Dictionary)

`dict` (từ điển) là cấu trúc dữ liệu cho phép liên kết **khoá (key)** với **giá trị (value)**. Trong khi `list` là danh sách các giá trị, `dict` liên kết mỗi khoá với đúng một giá trị.

Giả sử ta muốn gán mỗi học sinh Hogwarts vào một nhà (house). Ta **CÓ THỂ** dùng hai `list` riêng biệt:

````python
students = ["Hermione", "Harry", "Ron", "Draco"]
houses = ["Gryffindor", "Gryffindor", "Gryffindor", "Slytherin"]
````

Cách này yêu cầu ta phải "hứa" rằng hai `list` luôn được giữ **đúng thứ tự tương ứng**. Điều này khá rủi ro và cồng kềnh khi danh sách càng lớn!

Ta có cách tốt hơn — dùng `dict`:

````python
students = {
    "Hermione": "Gryffindor",
    "Harry": "Gryffindor",
    "Ron": "Gryffindor",
    "Draco": "Slytherin",
}

print(students["Hermione"])
print(students["Harry"])
print(students["Ron"])
print(students["Draco"])
````

Ta dùng `{}` (dấu ngoặc nhọn) để tạo `dict`. Nếu `list` dùng số (chỉ số) để truy cập phần tử, `dict` cho phép dùng **từ ngữ** (hoặc giá trị bất biến khác) làm "chỉ số" — gọi là **khoá (key)**.

Kết quả mong đợi:
```
Gryffindor
Gryffindor
Gryffindor
Slytherin
```

Ta cải thiện bằng vòng lặp:

````python
students = {
    "Hermione": "Gryffindor",
    "Harry": "Gryffindor",
    "Ron": "Gryffindor",
    "Draco": "Slytherin",
}

for student in students:
    print(student)
````

Vòng lặp `for` chỉ duyệt qua các **khoá (key)** của `dict` — kết quả chỉ là tên học sinh. Làm sao để in cả khoá **VÀ** giá trị?

````python
students = {
    "Hermione": "Gryffindor",
    "Harry": "Gryffindor",
    "Ron": "Gryffindor",
    "Draco": "Slytherin",
}

for student in students:
    print(student, students[student])
````

`students[student]` tra vào khoá `student` và lấy ra giá trị (nhà) tương ứng. Chạy thử — kết quả in ra hơi "lộn xộn" (các giá trị chỉ cách nhau khoảng trắng mặc định).

Ta dọn dẹp bằng tham số `sep` của `print()`:

````python
students = {
    "Hermione": "Gryffindor",
    "Harry": "Gryffindor",
    "Ron": "Gryffindor",
    "Draco": "Slytherin",
}

for student in students:
    print(student, students[student], sep=", ")
````

`sep=", "` (dấu phẩy và một khoảng trắng) được chèn giữa các giá trị mà `print` in ra. Kết quả:

## 🎮 Mario: Vòng lặp lồng nhau (Nested Loops)

Hẳn bạn còn nhớ trò chơi kinh điển **Mario**, nơi người hùng nhảy qua các viên gạch (bricks). Hãy tạo một phiên bản biểu diễn bằng văn bản — bài toán tuyệt vời để luyện **vòng lặp lồng nhau (nested loops)** kết hợp trừu tượng hoá (abstraction) bằng hàm.

Bắt đầu đơn giản — in một cột gồm 3 viên gạch:

````python
print("#")
print("#")
print("#")
````

Ta đang lặp lại chính đoạn code này — dấu hiệu rõ ràng cho thấy nên dùng vòng lặp:

````python
for _ in range(3):
    print("#")
````

Đoạn code đạt đúng hiệu quả mong muốn. Nhưng ta có thể **trừu tượng hoá thêm** để tái sử dụng cho các bài toán phức tạp hơn:

````python
def main():
    print_column(3)


def print_column(height):
    for _ in range(height):
        print("#")


main()
````

Giờ đây, cột gạch có thể "cao" bao nhiêu tuỳ ý, không cần sửa code cứng (hard-code).

Tiếp theo, hãy thử in một hàng ngang:

````python
def main():
    print_row(4)


def print_row(width):
    print("?" * width)


main()
````

Giờ ta có thể tạo các khối gạch từ trái sang phải.

Trong trò chơi Mario, các viên gạch không chỉ xếp thành một hàng hay một cột — mà tạo thành một khối vuông/chữ nhật gồm cả hàng **VÀ** cột. Làm sao kết hợp cả hai?

````python
def main():
    print_square(3)


def print_square(size):
    # For each row in square
    for i in range(size):
        # For each brick in row
        for j in range(size):
            # Print brick
            print("#", end="")
        # Print blank line
        print()


main()
````

Đây là một **vòng lặp lồng nhau (nested loop)**:
- Vòng lặp **ngoài** (biến `i`) chạy qua từng **hàng** của hình vuông.
- Với mỗi hàng, vòng lặp **trong** (biến `j`) in ra từng **viên gạch** (dùng `end=""` để các viên gạch nằm liền nhau trên cùng dòng).
- Sau khi vòng lặp trong hoàn tất một hàng, `print()` (không tham số) in ra dấu xuống dòng, chuẩn bị cho hàng kế tiếp.

Ta có thể trừu tượng hoá xa hơn, tái sử dụng luôn `print_row` đã viết ở trên:

````python
def main():
    print_square(3)


def print_square(size):
    for i in range(size):
        print_row(size)


def print_row(width):
    print("#" * width)


main()
````

Cách viết này ngắn gọn hơn, và **mỗi hàm chỉ đảm nhiệm một công việc rõ ràng, duy nhất** (`print_square` lo việc lặp qua các hàng; `print_row` lo việc in một hàng gạch) — một nguyên tắc thiết kế phần mềm quan trọng bạn sẽ còn gặp lại nhiều lần trong học phần.
