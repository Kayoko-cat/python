**Lưu ý khối lệnh:**

```python
if __name__ == "__main__":
    main()
```
Đây là “điểm vào” của chương trình. Khi bạn chạy file này trực tiếp (ví dụ: `python calculator.py`), biến đặc biệt `__name__` sẽ là `__main__`, nên `main()` được gọi.
Nếu file này được import từ một module khác, `main()` sẽ không tự động chạy. Điều này giúp tách biệt việc định nghĩa hàm và việc thực thi chương trình.

````python
from calculator import square


def main():
    test_square()


def test_square():
    try:
        assert square(2) == 4
    except AssertionError:
        print("2 squared is not 4")
    try:
        assert square(3) == 9
    except AssertionError:
        print("3 squared is not 9")
    try:
        assert square(-2) == 4
    except AssertionError:
        print("-2 squared is not 4")
    try:
        assert square(-3) == 9
    except AssertionError:
        print("-3 squared is not 9")
    try:
        assert square(0) == 0
    except AssertionError:
        print("0 squared is not 0")


if __name__ == "__main__":
    main()
````

````python
# test_calculator.py
# Lưu ý với pytest: Tên file kiểm thử phải bắt đầu bằng test_ hoặc kết thúc bằng _test.py
# Tên của các hàm kiểm thử phải bắt đầu bằng test_ để pytest có thể tự động phát hiện và
# chạy chúng. 
from calculator import square


def test_positive():
    assert square(2) == 4
    assert square(3) == 9


def test_negative():
    assert square(-2) == 4
    assert square(-3) == 9


def test_zero():
    assert square(0) == 0
````

````python
# test_calculator.py
import pytest

from calculator import square

def test_positive():
    assert square(2) == 4
    assert square(3) == 9


def test_negative():
    assert square(-2) == 4
    assert square(-3) == 9


def test_zero():
    assert square(0) == 0


def test_str():
    with pytest.raises(TypeError):
        square("cat")
````

**Hàm test mới: `test_str()`**

Mục đích: Đảm bảo hàm `square` xử lý (hoặc từ chối) đầu vào kiểu chuỗi đúng cách.
Nội dung:
* `with pytest.raises(TypeError)`:
    - Đây là cách viết trong `pytest` để khẳng định rằng khối lệnh bên trong sẽ phát sinh một ngoại lệ `TypeError`. Nếu ngoại lệ đó không xảy ra, hoặc xảy ra loại ngoại lệ khác, bài test sẽ thất bại.
* `square("cat")`
    - Gọi `square` với một chuỗi. Với hàm `square` hiện tại (`return n * n`), hành vi phụ thuộc vào kiểu `n`:
    - Với chuỗi `"cat"`, phép `"cat" * "cat"` là không hợp lệ trong Python, nên sẽ ném `TypeError`. Vì vậy test mong đợi `TypeError` là hợp lý.
    - Lưu ý: Nếu ai đó thay đổi `square` thành `n ** 2`, thì `"cat" ** 2` cũng sẽ ném `TypeError`. Còn trong trường hợp `n` là `list`, `tuple`, v.v., kiểu ngoại lệ có thể khác nhau theo phép toán.
