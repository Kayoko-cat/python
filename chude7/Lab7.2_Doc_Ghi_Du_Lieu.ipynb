## Giới thiệu

Cho đến nay, mọi chương trình chúng ta viết đều lưu trữ thông tin trong **bộ nhớ** (memory). Khi chương trình kết thúc, tất cả dữ liệu người dùng nhập vào hoặc do chương trình tạo ra sẽ bị mất.

**Đọc/ghi dữ liệu (File I/O – Input/Output)** là khả năng của một chương trình đọc dữ liệu từ một tệp tin (file) làm đầu vào, hoặc tạo ra một tệp tin làm đầu ra.

Để bắt đầu, hãy chạy đoạn code sau:

```python
name = input("What's your name? ")
print(f"hello, {name}")
```

Đoạn code này chạy đúng như mong đợi: người dùng nhập vào một tên, chương trình in ra lời chào.

Nhưng nếu ta muốn nhập vào **nhiều** tên? Hãy nhớ lại rằng `list` (danh sách) cho phép ta lưu nhiều giá trị vào một biến:

```python
names = []
for _ in range(3):
    name = input("What's your name? ")
    names.append(name)

print(names)
```

Người dùng sẽ được hỏi ba lần. Phương thức `append` được dùng để thêm `name` vào danh sách `names`.

Đoạn code trên có thể được viết gọn hơn:

```python
names = []
for _ in range(3):
    names.append(input("What's your name? "))

print(names)
```

Kết quả không thay đổi. Giờ, hãy in ra danh sách tên theo thứ tự đã sắp xếp:

```python
names = []
for _ in range(3):
    names.append(input("What's your name? "))

for name in sorted(names):
    print(f"hello, {name}")
```

Chú ý: **khi ô code (cell) chạy xong, toàn bộ thông tin sẽ bị mất** — chạy lại ô code trên, danh sách `names` sẽ trống trở lại. Đọc/ghi dữ liệu (File I/O) cho phép chương trình **lưu trữ** thông tin này để có thể sử dụng lại sau này, kể cả sau khi tắt máy.

> 📖 Tìm hiểu thêm: [Python docs — `sorted`](https://docs.python.org/3/library/functions.html#sorted)

## 7.1. Đọc/ghi file văn bản

### Hàm `open`

`open` là một hàm có sẵn trong Python, cho phép bạn mở một tệp tin và sử dụng nó trong chương trình — để đọc từ tệp hoặc ghi vào tệp.

Hãy quay lại và thử ghi tên vào một tệp văn bản:

```python
name = input("What's your name? ")

file = open("names.txt", "w")
file.write(name)
file.close()
```

Hàm `open` mở một tệp tin tên `names.txt` với chế độ **ghi** (write), được biểu thị bằng `"w"`. Đoạn code trên gán tệp đã mở cho biến `file`. Dòng `file.write(name)` ghi tên vào tệp văn bản; dòng kế tiếp **đóng** tệp lại.

Chạy ô code trên nhiều lần với các tên khác nhau, bạn sẽ thấy chương trình **ghi đè toàn bộ** tệp `names.txt` mỗi lần chạy — chỉ tên mới nhất được lưu lại.

Lý tưởng nhất, ta muốn **thêm (append)** từng tên vào tệp mà không mất tên cũ. Ta xoá tệp `names.txt` hiện có (nếu có), rồi thử lại với chế độ `"a"`:

```python
import os
if os.path.exists("names.txt"):
    os.remove("names.txt")

name = input("What's your name? ")

file = open("names.txt", "a")
file.write(name)
file.close()
```

Thay đổi duy nhất là `"w"` đã được đổi thành `"a"` (append – thêm vào). Chạy lại ô code này nhiều lần, bạn sẽ thấy các tên được thêm vào tệp. Tuy nhiên, kiểm tra nội dung tệp `names.txt` (ví dụ dùng lệnh `!cat names.txt` hoặc `!type names.txt` trên Windows), bạn sẽ thấy các tên bị **dính liền** vào nhau, không có khoảng cách.

Ta có thể khắc phục bằng cách thêm dấu xuống dòng `\n` sau mỗi tên. Xoá tệp `names.txt` rồi thử lại:

```python
import os
if os.path.exists("names.txt"):
    os.remove("names.txt")

name = input("What's your name? ")

file = open("names.txt", "a")
file.write(f"{name}\n")
file.close()
```

Dòng `file.write` đã được sửa để thêm một dấu xuống dòng (`\n`) ở cuối mỗi tên. Đoạn code này hoạt động khá tốt. Tuy nhiên, vẫn có cách để cải thiện — vì ta rất dễ **quên đóng tệp** (`file.close()`), điều này có thể khiến dữ liệu không được lưu đúng cách.

> 📖 Tìm hiểu thêm: [Python docs — `open`](https://docs.python.org/3/library/functions.html#open)

### Câu lệnh `with`

Từ khóa `with` cho phép bạn **tự động đóng tệp** sau khi sử dụng, ngay cả khi có lỗi xảy ra bên trong khối lệnh.

```python
name = input("What's your name? ")

with open("names.txt", "a") as file:
    file.write(f"{name}\n")
```

Chú ý rằng dòng bên dưới `with` được thụt vào (indented), và ta **không cần gọi `file.close()`** — Python sẽ tự động đóng tệp khi khối lệnh `with` kết thúc.

Từ đầu đến giờ, ta chỉ mới **ghi** vào tệp. Nếu ta muốn **đọc** từ tệp thì làm như thế nào?

```python
with open("names.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print("hello,", line)
```

`readlines` có khả năng đọc tất cả các dòng trong tệp và lưu chúng vào một danh sách gọi là `lines`. Chạy ô code này, bạn sẽ thấy kết quả in ra khá "xấu" — có những dấu xuống dòng thừa, vì mỗi `line` vẫn còn giữ ký tự `\n` ở cuối, và `print` lại tự thêm một dấu xuống dòng khác.

Ta có thể khắc phục bằng `rstrip`:

```python
with open("names.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print("hello,", line.rstrip())
```

`rstrip` có tác dụng loại bỏ dấu xuống dòng dư ở cuối mỗi dòng. Đoạn code này có thể đơn giản hoá hơn nữa — ta không cần `readlines()` vì bản thân `file` đã có thể lặp qua trực tiếp theo từng dòng:

```python
with open("names.txt", "r") as file:
    for line in file:
        print("hello,", line.rstrip())
```

Đoạn code này cho ra kết quả đúng. Tuy nhiên, ta chưa **sắp xếp** các tên. Ta có thể cải tiến bằng cách thu thập các tên vào một `list` trong bộ nhớ trước, sau đó sắp xếp và in ra:

```python
names = []
with open("names.txt") as file:
    for line in file:
        names.append(line.rstrip())

for name in sorted(names):
    print(f"hello, {name}")
```

`names` là một danh sách trống để ta thu thập các tên. Mỗi tên được `append` vào danh sách `names` trong bộ nhớ. Sau đó, mỗi tên trong danh sách đã sắp xếp được in ra. Chạy code, bạn sẽ thấy các tên đã được sắp xếp đúng thứ tự — và quan trọng hơn, **dữ liệu vẫn còn đó** dù bạn có tắt và mở lại notebook, vì nó đã được lưu trong tệp `names.txt` trên đĩa.

Nếu ta muốn lưu trữ nhiều hơn chỉ tên của sinh viên thì sao? Ví dụ, ta muốn lưu cả **tên** và **nhà** (house) của sinh viên đó? Đây là lúc ta cần đến các định dạng dữ liệu có cấu trúc hơn: **CSV**, **Excel**, và **SQL**.

## 7.2. Đọc/ghi dữ liệu CSV, Excel, SQL

### CSV

CSV là viết tắt của "comma-separated values" (các giá trị được phân tách bằng dấu phẩy) — một trong những định dạng phổ biến nhất để lưu dữ liệu dạng bảng.

Trong Jupyter Notebook, ta có thể dùng "magic command" `%%writefile` để tạo trực tiếp một tệp văn bản từ nội dung của một ô code (dòng `%%writefile ten_file` phải là dòng đầu tiên của ô):

```python
%%writefile students.csv
Hermione,Gryffindor
Harry,Gryffindor
Ron,Gryffindor
Draco,Slytherin
```

Bây giờ hãy đọc tệp `students.csv` vừa tạo:

```python
with open("students.csv") as file:
    for line in file:
        row = line.rstrip().split(",")
        print(f"{row[0]} is in {row[1]}")
```

`rstrip` loại bỏ dấu xuống dòng ở cuối mỗi dòng trong tệp CSV. `split` cho interpreter biết nơi kết thúc mỗi giá trị (dựa vào dấu phẩy). `row[0]` là phần tử đầu tiên của mỗi dòng, `row[1]` là phần tử thứ hai.

Đoạn code trên khá hiệu quả, nhưng cú pháp `row[0]`, `row[1]` khá khó hiểu nếu chưa quen. Python hỗ trợ cách viết đơn giản hơn, tận dụng việc `split` trả về nhiều giá trị cùng lúc:

```python
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        print(f"{name} is in {house}")
```

`split` thực chất trả về hai giá trị: giá trị trước dấu phẩy và giá trị sau dấu phẩy. Nhờ đó, ta có thể gán trực tiếp cho hai biến `name` và `house` cùng lúc!

Giả sử ta muốn in ra danh sách đã **sắp xếp**:

```python
students = []
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        students.append(f"{name} is in {house}")

for student in sorted(students):
    print(student)
```

Ta tạo một `list` gọi là `students`, `append` từng chuỗi vào danh sách này, rồi in ra danh sách đã sắp xếp.

Hãy nhớ lại rằng Python hỗ trợ `dictionary`, cho phép liên kết một khóa (key) với một giá trị (value). Ta có thể cải tiến code để lưu trữ dữ liệu có cấu trúc hơn (tách riêng `name` và `house` thay vì gộp chung thành một chuỗi):

```python
students = []
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {}
        student["name"] = name
        student["house"] = house
        students.append(student)

for student in students:
    print(f"{student['name']} is in {student['house']}")
```

Ta có thể viết gọn hơn bằng cách khai báo `dictionary` trực tiếp bằng cú pháp `{}`:

```python
students = []
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)

for student in students:
    print(f"{student['name']} is in {student['house']}")
```

Đoạn code này cho kết quả như mong muốn, chỉ thiếu việc **sắp xếp** sinh viên. Đáng tiếc, ta không thể `sorted(students)` trực tiếp như trước vì mỗi sinh viên bây giờ là một `dictionary` bên trong một `list` — Python không biết nên so sánh theo `name` hay `house`. Ta cần cho `sorted` biết cách lấy "khóa" (key) để so sánh, bằng cách viết một hàm riêng và truyền vào tham số `key`:

```python
students = []
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        students.append({"name": name, "house": house})

def get_name(student):
    return student["name"]

for student in sorted(students, key=get_name):
    print(f"{student['name']} is in {student['house']}")
```

Python hỗ trợ tham số `key` cho hàm `sorted`. Hàm `get_name` chỉ đơn giản trả về giá trị của `student["name"]`. Chạy đoạn code này, danh sách sẽ được sắp xếp theo tên.

Nếu một hàm như `get_name` chỉ được dùng đúng một lần, ta có thể viết gọn hơn bằng `lambda` — một hàm không có tên (anonymous function):

```python
students = []
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        students.append({"name": name, "house": house})

for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is in {student['house']}")
```

Cách đọc `lambda student: student["name"]` có thể diễn giải là: "Này Python, đây là một hàm không tên: cho một `student`, hãy lấy `name` của nó và trả về cho `key`."

Tuy nhiên, code của ta khá "mong manh" (fragile). Giả sử tệp CSV lưu **nơi ở** (home) thay vì nhà (house), và một trong các giá trị đó lại chứa dấu phẩy:

```python
%%writefile students.csv
Harry,"Number Four, Privet Drive"
Ron,The Burrow
Draco,Malfoy Manor
```

Thử chạy lại chương trình cũ (đã đổi `house` thành `home`) với tệp CSV mới này. Đoạn code dưới đây được đặt trong `try/except` để bạn thấy rõ lỗi phát sinh mà không làm dừng toàn bộ notebook:

```python
students = []
try:
    with open("students.csv") as file:
        for line in file:
            name, home = line.rstrip().split(",")
            students.append({"name": name, "home": home})

    for student in sorted(students, key=lambda student: student["name"]):
        print(f"{student['name']} is in {student['home']}")
except ValueError as e:
    print(f"Lỗi xảy ra: {e}")
```

Bạn sẽ thấy lỗi `ValueError: too many values to unpack`. Lỗi này xảy ra vì chương trình `split` dòng dữ liệu theo dấu phẩy `,`, nhưng chính giá trị "Number Four, Privet Drive" lại **chứa** một dấu phẩy bên trong nó — khiến `split(",")` trả về **3** phần tử thay vì 2, và Python không thể gán 3 giá trị cho 2 biến (`name`, `home`).

May mắn là đã có người phát triển sẵn cách để "phân tích" (parse) tệp CSV một cách chính xác: thư viện có sẵn `csv` của Python, với một đối tượng gọi là `reader`. `reader` hoạt động trong vòng lặp `for`, mỗi lần lặp cho ta một dòng (row) — bản chất là một `list`, trong đó `row[0]` là phần tử đầu tiên, `row[1]` là phần tử thứ hai:

```python
import csv

students = []
with open("students.csv") as file:
    reader = csv.reader(file)
    for row in reader:
        students.append({"name": row[0], "home": row[1]})

for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is from {student['home']}")
```

Chương trình giờ hoạt động đúng như mong đợi, dù có dấu phẩy thừa trong dữ liệu, vì `csv.reader` đủ "thông minh" để nhận biết dấu phẩy nào nằm trong dấu ngoặc kép (`"..."`) và không tách tại đó.

Cách thiết kế tốt hơn nữa là khai báo rõ ràng ngay trong tệp CSV bằng một dòng **tiêu đề (header)**, cho biết mỗi cột chứa loại dữ liệu gì:

```python
%%writefile students.csv
name,home
Harry,"Number Four, Privet Drive"
Ron,The Burrow
Draco,Malfoy Manor
```

Ta có thể dùng một phần của thư viện `csv` gọi là `DictReader` để xử lý tệp CSV linh hoạt hơn, dựa vào dòng tiêu đề này:

```python
import csv

students = []
with open("students.csv") as file:
    reader = csv.DictReader(file)
    for row in reader:
        students.append({"name": row["name"], "home": row["home"]})

for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is in {student['home']}")
```

`DictReader` trả về từng dòng dưới dạng một `dictionary`, cho phép truy cập trực tiếp bằng tên cột (`row["name"]`, `row["home"]`) thay vì phải nhớ vị trí `row[0]`, `row[1]`. Đây là một ví dụ về **lập trình phòng thủ** (coding defensively): chỉ cần người thiết kế tệp CSV nhập đúng dòng tiêu đề, chương trình vẫn hoạt động đúng dù thứ tự cột có thay đổi.

Từ đầu đến giờ, ta chỉ mới **đọc** tệp CSV. Nếu ta muốn **ghi** vào tệp CSV thì làm như thế nào? Hãy xoá tệp `students.csv` hiện có rồi thử ghi dữ liệu mới vào:

```python
import csv
import os

if os.path.exists("students.csv"):
    os.remove("students.csv")

name = input("What's your name? ")
home = input("Where's your home? ")

with open("students.csv", "a") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "home"])
    writer.writerow({"name": name, "home": home})
```

Ta tận dụng chức năng có sẵn của `DictWriter`, nhận hai tham số: `file` cần ghi vào và `fieldnames` (danh sách tên cột) cần ghi. Hàm `writerow` nhận một `dictionary` làm tham số — ta đang yêu cầu interpreter ghi một dòng gồm hai trường (field) gọi là `name` và `home`. Chạy lại ô code này nhiều lần với các giá trị khác nhau để thêm nhiều dòng vào `students.csv`.

#### Ghi dữ liệu vào tệp Excel

```python
import openpyxl

# Tạo một workbook (tệp Excel) mới và lấy sheet đang hoạt động
workbook = openpyxl.Workbook()
sheet = workbook.active

# Ghi dòng tiêu đề (header)
sheet.append(["name", "house"])

# Ghi dữ liệu từng sinh viên, mỗi sinh viên là một dòng (row)
students = [
    {"name": "Hermione", "house": "Gryffindor"},
    {"name": "Harry", "house": "Gryffindor"},
    {"name": "Ron", "house": "Gryffindor"},
    {"name": "Draco", "house": "Slytherin"},
]
for student in students:
    sheet.append([student["name"], student["house"]])

# Lưu workbook thành tệp .xlsx
workbook.save("students.xlsx")
print("Đã lưu students.xlsx")
```

`openpyxl.Workbook()` tạo ra một "quyển sổ tính" (workbook) mới trong bộ nhớ, giống như khi bạn mở Excel và tạo một file trống. `workbook.active` cho ta "trang tính" (sheet) hiện đang hoạt động. Phương thức `sheet.append(...)` thêm một dòng mới vào cuối sheet — hoạt động rất giống `list.append` mà ta đã quen! Cuối cùng, `workbook.save(...)` ghi toàn bộ nội dung ra tệp `.xlsx`.

#### Đọc dữ liệu từ tệp Excel

```python
import openpyxl

workbook = openpyxl.load_workbook("students.xlsx")
sheet = workbook.active

students = []
for row in sheet.iter_rows(min_row=2, values_only=True):  # bỏ dòng tiêu đề
    name, house = row
    students.append({"name": name, "house": house})

for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is in {student['house']}")
```

`load_workbook` mở tệp `.xlsx` đã tồn tại. `sheet.iter_rows(...)` cho ta lặp qua từng dòng của sheet — tương tự việc lặp qua từng `line` trong tệp văn bản. Tham số `min_row=2` giúp bỏ qua dòng tiêu đề, và `values_only=True` giúp lấy trực tiếp giá trị của các ô (cell) thay vì đối tượng `Cell`. Phần còn lại (thêm vào `list`, sắp xếp, in ra) hoàn toàn giống như những gì ta đã làm với CSV.

### SQL

Một cách khác — và thường mạnh mẽ hơn khi dữ liệu lớn — để lưu trữ dữ liệu có cấu trúc là sử dụng **cơ sở dữ liệu** (database) thông qua **SQL** (Structured Query Language). Python có sẵn thư viện `sqlite3` để làm việc với **SQLite** — một loại cơ sở dữ liệu SQL nhẹ, lưu toàn bộ dữ liệu trong **một tệp duy nhất** (ví dụ `students.db`), không cần cài đặt phần mềm máy chủ (server) nào thêm.

#### Tạo cơ sở dữ liệu và bảng dữ liệu

```python
import sqlite3

# Kết nối tới tệp cơ sở dữ liệu (sẽ được tạo mới nếu chưa tồn tại)
connection = sqlite3.connect("students.db")
cursor = connection.cursor()

# Tạo bảng "students" nếu chưa có
cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        name TEXT,
        house TEXT
    )
""")

connection.commit()
connection.close()
print("Đã tạo bảng students trong students.db")
```

`sqlite3.connect("students.db")` mở (hoặc tạo mới) tệp cơ sở dữ liệu. `cursor` giống như "con trỏ" cho phép thực thi các câu lệnh SQL. `cursor.execute(...)` thực thi câu lệnh `CREATE TABLE`, tạo một **bảng** (table) tên `students` với hai **cột** (column) `name` và `house`, kiểu văn bản (`TEXT`). `connection.commit()` yêu cầu SQLite thực sự lưu các thay đổi vào tệp. `connection.close()` đóng kết nối — **khác với tệp văn bản**, `sqlite3` không tự đóng kết nối, nên ta luôn cần tự gọi `close()`.

#### Ghi (thêm) dữ liệu vào cơ sở dữ liệu

```python
import sqlite3

name = input("What's your name? ")
house = input("What's your house? ")

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("INSERT INTO students (name, house) VALUES (?, ?)", (name, house))

connection.commit()
connection.close()
print("Đã thêm dữ liệu vào students.db")
```

Hai dấu `?` là các "chỗ giữ trống" (placeholder), được thay bằng các giá trị trong tuple `(name, house)`. Đây gọi là **truy vấn có tham số** (parameterized query) — an toàn hơn nhiều so với việc nối chuỗi trực tiếp (ví dụ `f"INSERT INTO students VALUES ('{name}', '{house}')"`), vì giúp chống lại lỗ hổng bảo mật phổ biến gọi là **SQL injection**.

Chạy lại ô code trên vài lần với các giá trị khác nhau để có nhiều dòng dữ liệu.

#### Đọc (truy vấn) dữ liệu từ cơ sở dữ liệu

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("SELECT name, house FROM students")
rows = cursor.fetchall()

students = []
for name, house in rows:
    students.append({"name": name, "house": house})

connection.close()

for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is in {student['house']}")
```

`SELECT name, house FROM students` yêu cầu cơ sở dữ liệu trả về hai cột đó của tất cả các dòng trong bảng. `cursor.fetchall()` lấy toàn bộ kết quả về dưới dạng một `list` các `tuple`. Phần còn lại — thêm vào `list`, sắp xếp bằng `key=lambda`, in ra — hoàn toàn giống cách ta đã xử lý với CSV và Excel. Dù dữ liệu đến từ nguồn nào, một khi đã đưa vào các cấu trúc dữ liệu quen thuộc của Python (`list`, `dictionary`), cách xử lý tiếp theo đều tương tự nhau.

> ⚠️ **Lưu ý:** bạn cũng có thể dùng cú pháp `with sqlite3.connect("students.db") as connection:`. Tuy nhiên, khác với `open`, câu lệnh `with` ở đây **chỉ tự động `commit()` (nếu thành công) hoặc `rollback()` (nếu có lỗi)** — nó **không tự động đóng kết nối**. Bạn vẫn cần tự gọi `connection.close()` khi dùng xong.

> 📖 Tìm hiểu thêm: [Python docs — `sqlite3`](https://docs.python.org/3/library/sqlite3.html)

### So sánh nhanh: khi nào dùng định dạng nào?

| Định dạng | Ưu điểm | Phù hợp khi |
|---|---|---|
| **TXT** | Đơn giản nhất, không cấu trúc | Ghi log, danh sách đơn giản |
| **CSV** | Nhẹ, dễ đọc bằng mắt, tương thích Excel/Google Sheets | Chia sẻ dữ liệu dạng bảng đơn giản |
| **Excel (.xlsx)** | Người dùng phi kỹ thuật dễ mở/sửa, hỗ trợ nhiều sheet | Báo cáo, làm việc với dân văn phòng |
| **SQL (SQLite)** | Truy vấn nhanh, đảm bảo tính nhất quán, xử lý dữ liệu lớn | Ứng dụng cần tìm/lọc/cập nhật dữ liệu thường xuyên |

> 💡 Lưu ý: thư viện **Pandas** (sẽ học ở Chủ đề 8) hỗ trợ đọc/ghi trực tiếp cả CSV, Excel và SQL chỉ với một dòng lệnh như `pandas.read_csv()`, `pandas.read_excel()`, `pandas.read_sql()`, giúp đơn giản hoá rất nhiều bước ta vừa thực hiện thủ công ở trên.

## Tổng kết

Trong chủ đề này, ta đã học được rằng một chương trình không chỉ có thể lưu trữ dữ liệu tạm thời trong bộ nhớ, mà còn có thể **đọc** dữ liệu từ tệp tin (hoặc cơ sở dữ liệu) và **ghi** dữ liệu ra để sử dụng lại sau này:

- **Tệp văn bản (`.txt`)**: dùng hàm `open` với các chế độ `"r"` (đọc), `"w"` (ghi, ghi đè), `"a"` (ghi, thêm vào cuối); nên dùng `with` để tự động đóng tệp.
- **Tệp CSV (`.csv`)**: có thể tự `split` từng dòng theo dấu phẩy, hoặc dùng thư viện có sẵn `csv` với `reader`/`writer` (đọc/ghi theo danh sách) và `DictReader`/`DictWriter` (đọc/ghi theo dictionary, dựa vào dòng tiêu đề).
- **Tệp Excel (`.xlsx`)**: dùng thư viện bên thứ ba `openpyxl`, với `Workbook`/`load_workbook` để tạo/mở tệp, `sheet.append` để ghi dòng, và `sheet.iter_rows` để đọc dòng.
- **Cơ sở dữ liệu SQL (`.db`)**: dùng thư viện có sẵn `sqlite3`, với `connect`/`cursor` để kết nối, `execute` để thực thi câu lệnh SQL (`CREATE TABLE`, `INSERT`, `SELECT`,...), `fetchall` để lấy kết quả, và luôn `commit`/`close` kết nối khi hoàn tất.

Dù dữ liệu được lưu ở định dạng nào, mục tiêu cuối cùng vẫn là đưa dữ liệu đó vào các cấu trúc dữ liệu quen thuộc của Python (`list`, `dictionary`) để có thể xử lý, sắp xếp, lọc và hiển thị. Đây chính là nền tảng quan trọng để bước sang các chủ đề tiếp theo, khi ta sử dụng các thư viện mạnh mẽ hơn như **NumPy** và **Pandas** để xử lý dữ liệu lớn một cách hiệu quả hơn.
