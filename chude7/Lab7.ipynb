## 1. Giới thiệu về File I/O

**File I/O** là khả năng của chương trình để:
- Đọc dữ liệu từ file (Input)
- Ghi dữ liệu ra file (Output)

Điều này giúp chúng ta lưu trữ dữ liệu lâu dài, không bị mất khi chương trình kết thúc.

## 2. Ghi File - Sử dụng `open()` function

### 2.1. Ghi file cơ bản

```python
# Ví dụ 1: Ghi tên vào file (mode 'w' - write, sẽ ghi đè file cũ)
name = input("What's your name? ")

file = open("names.txt", "w")
file.write(name)
file.close()

print("Đã ghi tên vào file names.txt")
```

**Vấn đề**: Code trên sẽ **ghi đè** (overwrite) file mỗi lần chạy. Để thêm dữ liệu vào cuối file, ta dùng mode `'a'` (append).

```python
# Ví dụ 2: Thêm tên vào file (mode 'a' - append)
name = input("What's your name? ")

file = open("names.txt", "a")
file.write(f"{name}\n")  # Thêm \n để xuống dòng
file.close()

print("Đã thêm tên vào file names.txt")
```

## 3. Sử dụng `with` statement

Từ khóa `with` giúp tự động đóng file, tránh quên gọi `file.close()`

```python
# Ví dụ 3: Sử dụng with để ghi file
name = input("What's your name? ")

with open("names.txt", "a") as file:
    file.write(f"{name}\n")

# File tự động được đóng sau khối with
print("Đã ghi file thành công!")
```

## 4. Đọc File

### 4.1. Đọc toàn bộ file với `readlines()`

```python
# Ví dụ 4: Đọc file names.txt
with open("names.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print("hello,", line.rstrip())  # rstrip() loại bỏ \n ở cuối
```

### 4.2. Đọc từng dòng một (hiệu quả hơn)

```python
# Ví dụ 5: Đọc và sắp xếp tên
names = []

with open("names.txt", "r") as file:
    for line in file:
        names.append(line.rstrip())

# Sắp xếp và in ra
for name in sorted(names):
    print(f"hello, {name}")
```

## 5. Làm việc với CSV Files

**CSV** (Comma Separated Values) là định dạng file phổ biến để lưu dữ liệu dạng bảng.

### 5.1. Tạo file CSV mẫu

```python
# Tạo file students.csv
with open("students.csv", "w") as file:
    file.write("Hermione,Gryffindor\n")
    file.write("Harry,Gryffindor\n")
    file.write("Ron,Gryffindor\n")
    file.write("Draco,Slytherin\n")

print("Đã tạo file students.csv")
```

### 5.2. Đọc CSV không dùng thư viện

```python
# Ví dụ 6: Đọc CSV bằng split()
with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        print(f"{name} is in {house}")
```

### 5.3. Lưu vào dictionary và sắp xếp

```python
# Ví dụ 7: Sử dụng dictionary
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        students.append({"name": name, "house": house})

# Sắp xếp theo tên sử dụng lambda function
for student in sorted(students, key=lambda student: student["name"]):
    print(f"{student['name']} is in {student['house']}")
```

## 6. Sử dụng thư viện `csv`

Python có thư viện `csv` tích hợp sẵn để xử lý CSV files hiệu quả hơn.

### 6.1. Tạo file CSV với header

```python
# Tạo file với header
with open("students.csv", "w") as file:
    file.write("name,home\n")
    file.write('Harry,"Number Four, Privet Drive"\n')
    file.write("Ron,The Burrow\n")
    file.write("Draco,Malfoy Manor\n")

print("Đã tạo file students.csv với header")
```

### 6.2. Đọc CSV với `csv.reader()`

```python
# Ví dụ 8: Sử dụng csv.reader
import csv

students = []

with open("students.csv") as file:
    reader = csv.reader(file)
    next(reader)  # Bỏ qua dòng header
    
    for row in reader:
        students.append({"name": row[0], "home": row[1]})

for student in sorted(students, key=lambda s: s["name"]):
    print(f"{student['name']} is from {student['home']}")
```

### 6.3. Đọc CSV với `csv.DictReader()` (Khuyên dùng)

```python
# Ví dụ 9: Sử dụng csv.DictReader - Tự động xử lý header
import csv

students = []

with open("students.csv") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        students.append({"name": row["name"], "home": row["home"]})

for student in sorted(students, key=lambda s: s["name"]):
    print(f"{student['name']} is from {student['home']}")
```

### 6.4. Ghi CSV với `csv.DictWriter()`

```python
# Ví dụ 10: Ghi dữ liệu vào CSV
import csv

name = input("What's your name? ")
home = input("Where's your home? ")

with open("students.csv", "a", newline='') as file:
    writer = csv.DictWriter(file, fieldnames=["name", "home"])
    writer.writerow({"name": name, "home": home})

print("Đã thêm sinh viên vào file!")
```

### 7.2. Tạo ảnh đơn giản

```python
# Ví dụ 11: Tạo ảnh màu đỏ
from PIL import Image

# Tạo ảnh 200x200 pixels, màu đỏ (RGB: 255, 0, 0)
img = Image.new('RGB', (200, 200), color='red')
img.save('red_square.png')

print("Đã tạo ảnh red_square.png")

# Hiển thị ảnh (nếu chạy trên máy local)
# img.show()
```

### 7.3. Mở và resize ảnh

```python
# Ví dụ 12: Mở và thay đổi kích thước ảnh
from PIL import Image

try:
    # Mở ảnh (thay 'your_image.jpg' bằng đường dẫn ảnh của bạn)
    img = Image.open('red_square.png')
    
    # Hiển thị thông tin
    print(f"Kích thước gốc: {img.size}")
    print(f"Định dạng: {img.format}")
    
    # Resize ảnh
    new_img = img.resize((100, 100))
    new_img.save('resized_image.png')
    
    print(f"Kích thước mới: {new_img.size}")
    print("Đã lưu ảnh resized_image.png")
    
except FileNotFoundError:
    print("Không tìm thấy file ảnh!")
```

### 7.4. Tạo GIF animation (Ví dụ từ bài giảng)

```python
# Ví dụ 13: Tạo GIF đơn giản
from PIL import Image, ImageDraw

# Tạo 2 frames cho animation
frames = []

# Frame 1: Hình tròn ở trái
img1 = Image.new('RGB', (400, 200), color='white')
draw1 = ImageDraw.Draw(img1)
draw1.ellipse([50, 50, 150, 150], fill='blue')
frames.append(img1)

# Frame 2: Hình tròn ở phải
img2 = Image.new('RGB', (400, 200), color='white')
draw2 = ImageDraw.Draw(img2)
draw2.ellipse([250, 50, 350, 150], fill='blue')
frames.append(img2)

# Lưu thành GIF
frames[0].save(
    'animation.gif',
    save_all=True,
    append_images=[frames[1]],
    duration=500,  # 500ms mỗi frame
    loop=0  # Lặp vô hạn
)

print("Đã tạo animation.gif")
```
