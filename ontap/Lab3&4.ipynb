## PHẦN 1: VÒNG LẶP WHILE

Vòng lặp while cho phép thực hiện một khối code nhiều lần cho đến khi điều kiện không còn đúng.

````python
# Ví dụ 1: Vòng lặp while cơ bản
print("=== Ví dụ vòng lặp while ===")
i = 0
while i < 3:
    print("meow")
    i += 1

print(f"\nGiá trị cuối cùng của i: {i}")
````

## PHẦN 2: VÒNG LẶP FOR

Vòng lặp for thường được sử dụng khi bạn biết trước số lần lặp hoặc muốn duyệt qua các phần tử trong một collection.

## PHẦN 3: CONTINUE VÀ BREAK

- `break`: Thoát khỏi vòng lặp ngay lập tức
- `continue`: Bỏ qua phần còn lại của iteration hiện tại và chuyển sang iteration tiếp theo

````python
# Ví dụ 4: Sử dụng break và continue
print("=== Ví dụ break ===")
for i in range(10):
    if i == 5:
        break
    print(i)

print("\n=== Ví dụ continue ===")
for i in range(10):
    if i % 2 == 0:  # Bỏ qua số chẵn
        continue
    print(i)
````

### 📝 BÀI TẬP 3: Break và Continue

**Yêu cầu:** Viết vòng lặp từ 1 đến 20:
- Bỏ qua các số chia hết cho 3 (dùng continue)
- Dừng lại khi gặp số 15 (dùng break)

````python
# BÀI LÀM CỦA BẠN:
# Viết code ở đây
print("== continue ==")
for i in range(1,21):
    if i % 3 == 0: # bỏ qua các số chia hết cho 3
        continue
    print(i,end=" ")

print("\n== break ==")
for i in range (1,21):
    if i == 15: # dừng lại khi gặp số 15
        break
    print(i,end=" ")
````

## 7.2. TRY-EXCEPT

Sử dụng try-except để bắt và xử lý lỗi

````python
# Ví dụ 10: Try-except cơ bản
def get_number_v1():
    try:
        x = int(input("What's x? "))
    except ValueError:
        print("x is not an integer")
        return None
    else:
        return x

# Test:
# result = get_number_v1()
# if result is not None:
#     print(f"x is {result}")
````

````python
# Ví dụ 11: Try-except với vòng lặp
def get_number_v2():
    while True:
        try:
            x = int(input("What's x? "))
        except ValueError:
            print("x is not an integer")
        else:
            return x

# Test:
# x = get_number_v2()
# print(f"x is {x}")
````

````python
# Ví dụ 12: Version tối ưu với pass
def get_int(prompt="Enter a number: "):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            pass  # Im lặng, chỉ hỏi lại

# Test:
x = get_int("What's x? ")
print(f"x is {x}")
````

### 📝 BÀI TẬP 7: Xử lý lỗi với try-except

**Yêu cầu:** Viết hàm `get_positive_int(prompt)` để:
1. Nhận input từ người dùng
2. Kiểm tra xem có phải số nguyên không (dùng try-except)
3. Kiểm tra xem số đó có dương không (> 0)
4. Nếu không hợp lệ, yêu cầu nhập lại
5. Trả về số nguyên dương hợp lệ

````python
# BÀI LÀM CỦA BẠN:
def get_positive_int(prompt="Enter a positive number: "):
    # Viết code ở đây
    while True:
        try:
            num =int(input(prompt))
            if num > 0:
                return num
            else:
                print("phải là số lớn hơn ko vui lòng nhập lại ")
        except ValueError:
            pass
# Test hàm của bạn:
 
 
n = get_positive_int("Enter a positive integer: ")
print(f"You entered: {n}")
````

## PHẦN 8: BÀI TẬP XỬ LÝ DỮ LIỆU

Kết hợp nhiều kỹ năng đã học

````python
# Ví dụ 13: Quản lý danh sách sinh viên
def student_manager():
    students = []
    
    while True:
        print("\n=== QUẢN LÝ SINH VIÊN ===")
        print("1. Thêm sinh viên")
        print("2. Hiển thị danh sách")
        print("3. Thoát")
        
        try:
            choice = int(input("Chọn (1-3): "))
        except ValueError:
            print("Vui lòng nhập số!")
            continue
        
        if choice == 1:
            name = input("Tên sinh viên: ")
            students.append(name)
            print(f"Đã thêm {name}")
        elif choice == 2:
            if not students:
                print("Danh sách trống!")
            else:
                print("\nDanh sách sinh viên:")
                for i, student in enumerate(students, 1):
                    print(f"{i}. {student}")
        elif choice == 3:
            print("Tạm biệt!")
            break
        else:
            print("Lựa chọn không hợp lệ!")

# Uncomment để chạy:
# student_manager()
````

### 📝 BÀI TẬP 8: Quản lý điểm số

**Yêu cầu:** Viết chương trình quản lý điểm với các chức năng:
1. Thêm sinh viên và điểm (lưu trong dictionary)
2. Hiển thị danh sách sinh viên và điểm
3. Tính điểm trung bình của lớp
4. Tìm sinh viên có điểm cao nhất
5. Xử lý lỗi khi nhập điểm (phải là số từ 0-10)

````python
# BÀI LÀM CỦA BẠN:
def grade_manager():
    
    grades = {}
    
    while True:
        print("\n=== QUẢN LÝ ĐIỂM ===")
        print("1. Thêm sinh viên và điểm")
        print("2. Hiển thị danh sách")
        print("3. Tính điểm trung bình")
        print("4. Tìm điểm cao nhất")
        print("5. Thoát")
        
        try:
            choice = int(input("Chọn (1-5): "))
        except ValueError:
            print("Vui lòng nhập số!")
            continue
        
        if choice == 1:
            name = input("Tên sinh viên: ")
            while True:
                try:
                    grade = float(input("Điểm (0-10): "))
                    if 0 <= grade <= 10:
                        grades[name] = grade
                        print(f"Đã thêm {name} với điểm {grade}")
                        break
                    else:
                        print("Điểm phải từ 0-10!")
                except ValueError:
                    print("Vui lòng nhập số!")
        
        elif choice == 2:
            if not grades:
                print("Chưa có sinh viên nào!")
            else:
                print("\nDanh sách:")
                for name, grade in grades.items():
                    print(f"{name}: {grade}")
        
        elif choice == 3:
            if not grades:
                print("Chưa có dữ liệu!")
            else:
                avg = sum(grades.values()) / len(grades)
                print(f"Điểm trung bình: {avg:.2f}")
        
        elif choice == 4:
            if not grades:
                print("Chưa có dữ liệu!")
            else:
                top_student = max(grades, key=grades.get)
                print(f"Điểm cao nhất: {top_student} - {grades[top_student]}")
        
        elif choice == 5:
            print("Tạm biệt!")
            break
        
        else:
            print("Lựa chọn không hợp lệ!")


grade_manager()

````

## TÓM TẮT KIẾN THỨC

### Loops:
- **while loop**: Lặp khi điều kiện đúng
- **for loop**: Lặp qua các phần tử hoặc range
- **break**: Thoát vòng lặp
- **continue**: Bỏ qua iteration hiện tại

### Data Structures:
- **list**: Danh sách có thứ tự, truy cập bằng index
- **dict**: Cặp key-value, truy cập bằng key
- **len()**: Độ dài của list/dict

### Exception Handling:
- **try-except**: Bắt và xử lý lỗi
- **else**: Chạy khi không có lỗi
- **pass**: Bỏ qua không làm gì
- **ValueError**: Lỗi giá trị không hợp lệ

### Best Practices:
1. Luôn validate user input
2. Sử dụng try-except cho code có thể gây lỗi
3. Đặt tên biến có ý nghĩa
4. Chia nhỏ code thành functions
5. Comment code khi cần thiết
